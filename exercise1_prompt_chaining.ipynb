# Exercise 1 - Prompt Chaining
Tools: Google Colab, Python, and an LLM API.

Prompts:
1. Classify the issue as billing, technical, shipping, cancellation, or general. Return a category and reason.
2. Using the category and original message, list only non-sensitive information needed. Never request passwords or full card numbers.
3. Using the category, missing information, and sentiment, draft a concise empathetic response. Do not promise unsupported refunds.
4. Escalate duplicate-charge billing cases; otherwise use standard support.

Iteration: the first draft requested sensitive account credentials and promised an immediate refund. The revised prompts prohibit sensitive data and unsupported promises.

```python
message = 'I was charged twice for my subscription.'
category = 'billing' if 'charged twice' in message else 'general'
missing = ['transaction date', 'amount', 'invoice number']
response = 'Thank you for reporting this billing issue. Please provide the transaction date, amount, and invoice number. Do not send payment-card numbers.'
route = 'Escalate to billing review' if category == 'billing' else 'Standard support'
print('STEP 1:', category, '- duplicate subscription charge')
print('STEP 2:', ', '.join(missing))
print('STEP 3:', response)
print('STEP 4:', route)
```
