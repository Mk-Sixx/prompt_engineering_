# Exercise 2 - ReAct Code Generation
Tools: Google Colab and Python.

Full ReAct prompt: Solve the task using this cycle: (1) Reason/Plan inputs, outputs, algorithm, and edge cases in no more than five bullets; (2) Generate Python using only the standard library; (3) Run a normal case and an edge case; (4) Observe errors or missing requirements; (5) Fix the code; (6) Run again. Require two-decimal output, reject negative prices and discounts outside 0-100 percent with ValueError, and show final output.

Iteration note: the first attempt worked for a normal input but lacked validation. The second attempt adds both validation rules.

```python
# Plan
print('PLAN: price and discount -> rounded final price; validate inputs')

# First generated attempt
def first_attempt(price, discount):
    return round(price * (1 - discount / 100), 2)
print('FIRST ATTEMPT:', first_attempt(100, 15))
print('OBSERVE: validation missing')

# Fixed generated code
def discounted_price(price, discount):
    if price < 0:
        raise ValueError('price cannot be negative')
    if not 0 <= discount <= 100:
        raise ValueError('discount must be between 0 and 100')
    return round(price * (1 - discount / 100), 2)
print('FIX: added checks for price and discount range')
print('FINAL:', discounted_price(100, 15))
try:
    discounted_price(100, 125)
except ValueError:
    print('EDGE CASE: ValueError handled successfully')
```
